```python
from ucimlrepo import fetch_ucirepo
import pandas as pd

student_performance = fetch_ucirepo(id=320)

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)

print(list(student_performance.data.features))

student_data = student_performance.data.features;
```

```python
import pandas as pd

# Beispiel-Daten
data = pd.DataFrame({
    'Grade': ['A', 'B', 'C', 'D', 'F']
})

# Reihenfolge der Grades definieren
grade_order = ['A', 'B', 'C', 'D', 'F']

# In ordinale Daten umwandeln
data['Grade'] = pd.Categorical(data['Grade'], categories=grade_order, ordered=True)

# Optional: In Zahlen umwandeln
data['Grade_Numeric'] = data['Grade'].cat.codes + 1

print(data)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Abwesenheitsdaten
absence = student_data['absences']

# Mean, Median und Modus berechnen
mean_absence = absence.mean()
median_absence = absence.median()
mode_absence = absence.mode()[0]

# Ergebnisse ausgeben
print(f"Mean: {mean_absence}")
print(f"Median: {median_absence}")
print(f"Modus: {mode_absence}")

# Verteilung plotten
plt.hist(absence, bins=20, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(mean_absence, color='red', linestyle='solid', linewidth=4, label=f'Mean: {mean_absence:.2f}')
plt.axvline(median_absence, color='green', linestyle='solid', linewidth=4, label=f'Median: {median_absence:.2f}')
plt.axvline(mode_absence, color='orange', linestyle='solid', linewidth=4, label=f'Modus: {mode_absence}')
plt.title('Verteilung der Abwesenheiten')
plt.xlabel('Anzahl der Abwesenheiten')
plt.ylabel('Häufigkeit')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
